puedes generar un programa en python que simule inteligencia artificial utilizando if. if else, elif, y while, referido a modelo y fotografo tipo menu, que sea interactivo.
```python
print("--- Asistente de Fotografía IA ---")
opcion = ""
while opcion != "3":
    print("\n1. Sugerir Pose\n2. Configurar Luz\n3. Finalizar")
    opcion = input("Selecciona una opción: ")
    if opcion == "1":
        estilo = input("¿Quieres algo 'urbano' o 'formal'?: ")
        if estilo == "urbano":
            print("IA: Sugerencia -> Sentada en las escaleras, mirada casual.")
        elif estilo == "formal":
            print("IA: Sugerencia -> De pie, manos a la cintura, espalda recta.")
        else:
            print("IA: Estilo no reconocido, intenta de nuevo.")
    elif opcion == "2":
        clima = input("¿Está 'soleado' o 'nublado'?: ")
        if clima == "soleado":
            print("IA: Recomendación -> Usa un filtro ND y busca sombras.")
        else:
            print("IA: Recomendación -> Sube el ISO y usa luz artificial.")
```
